```python
from dpvo import fastba
import torch
import sys
import os
sys.path.append('/home/campus.ncl.ac.uk/c4071391/Projects/DPVO/dpvo')

import onnx_mods

# edges defined as source -> patch -> target
source_frames = torch.tensor([0,0,0,1,1,1,2,2,2])
patches =       torch.tensor([1,2,3,1,2,3,1,2,3])
target_frames = torch.tensor([1,1,1,2,2,2,3,3,3])

# ix is the previous neighbour edge
# jx is the following neighbour edge
ix_1, jx_1 = fastba.neighbors(patches, target_frames)

print(f'----------fastba results----------')
print(f'ix: {ix_1}')
print(f'jx: {jx_1}')

ix_2, jx_2 = onnx_mods.neighbors(patches, target_frames)

print(f'----------onnx_mods results----------')
print(f'ix: {ix_2}')
print(f'jx: {jx_2}')

if not torch.equal(ix_1.cpu(), ix_2.cpu()) or not torch.equal(jx_1.cpu(), jx_2.cpu()):
    print(f'ERROR: onnx_mods method does not result in the same results as original method')

# at time of last testing it is just the edge cases of no previous or post neighbours found that messes up. 
```
